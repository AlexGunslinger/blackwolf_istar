```python
with open('D:\REGISTRATION DATA\Test-GUI\JB-641 21ft Cylinder Column/bottom/fine_alignment/fine_alignment_info.txt', 'r') as f:
    info = json.load(f)
```

```python
trans = np.array(list(info.values())[0])
```

```python
i = 0
s = bundle.stations[i]
s.ptc = s.estimate_scanner_position().reshape([1,3])
s.transform(
    s.align['rotation'],
    s.align['translation']
)
trans = np.array(list(info.values())[i])
s.transform(
    trans[:3, :3],
    trans[-1, :3].reshape([3,1])
)
A = s.ptc
```

```python
z_tops = np.array([s.z_top for s in bundle.stations if s.z_top < 3])
height = z_tops.mean()
height
```

```python
top_ptc = s.ptc[s.ptc[:, -1] > height + 1]
```

```python
B = top_ptc
```

```python
A[-1] = 1
```

```python
m = (A[0, -1] - height) / (A[0, -1] - B[:, -1])
```

```python
m = np.repeat(m.reshape([-1, 1]), 3, axis=1)
m.shape
```

```python
C = A - m * (A - B)
```

```python
C.mean(axis=0)
```
