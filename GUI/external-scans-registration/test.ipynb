## Get Job Info  
GET Domain/public_api.php?action=get_job_info&id={id}  
Here the id is not Job ID. The id is the unique id by job processing  
Example:  
GET https://ec2-3-141-161-185.us-east-2.compute.amazonaws.com/public_api.php?action=get_job_info&id=11

```python
params = dict(
    action='get_job_info',
    id=11
)
res = requests.get(uri, params, verify=False)
res
```

## Update Status  
GET Domain/public_api.php?action=update_status&id={id}&status={status}&message={message}  
Here the id is unique id of job. Status and message is string for job status and error or notification message

```python
params = dict(
    action='update_status',
    id='12',
    status='test api 2',
    # message='changing job status with api'
)
res = requests.get(uri, params, verify=False)
res.text
```
