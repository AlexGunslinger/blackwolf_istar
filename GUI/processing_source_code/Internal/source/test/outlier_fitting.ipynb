# Read transform matrices from running 2D alignment

```python
file_path = 'D:/REGISTRATION DATA/External/JB-595 117ft EXT - FARO/2D_aligned/transform_2d.pkl'
with open(file_path, 'rb') as f:
    transform_matrix = pickle.load(f)
```

```python
# data_dir = Path('D:/REGISTRATION DATA/External/JB-595 117ft EXT - FARO/tank_1/outliers')
data_dir = Path('D:/REGISTRATION DATA/External/JB-595 117ft EXT - FARO/tank_1')
write_dir = os.path.join(data_dir, 'align_2d')
```

```python
station_names = list(range(1,17)) + [23, 24, 25, 26]
if not os.path.exists(write_dir):
    os.mkdir(write_dir)
for sn,transf in zip(station_names, transform_matrix):
    c = Cylinder(
        metadata={
            'ptc_filepath': os.path.join(data_dir, '{:03d}.txt'.format(sn))
        }
    )
    print(c.ptc_filepath)
    c.read_points()
    c.transform(transf['rotation'], transf['translation'], update_axis=False)
    write_array2txt(
        os.path.join(write_dir, '{:03d}.txt'.format(sn)),
        c.points
    )
    del c
```
