```python

threshold  = 1 # folder 055, threshold = 1, folder 055_2, threshold = 0.5
# voxel_size = 0.02
# radius     = 0.01
# max_nn     = 20

trans_init = np.identity(4)

path = 'D:/REGISTRATION DATA/External/JB-595 117ft EXT - FARO/tank_1/outliers'
out_dir = path
if not os.path.exists(out_dir):    
    os.makedirs(out_dir,exist_ok=True)  

station_names = list(range(1,17)) + [23, 24, 25, 26]
with o3d.utility.VerbosityContextManager(
        o3d.utility.VerbosityLevel.Debug) as cm:
    for ia, a in enumerate(station_names):
        target = o3d.io.read_point_cloud(os.path.join(path,'{:03d}.txt'.format(a)),format='xyz')
        # target = target.voxel_down_sample(voxel_size)
        # target.estimate_normals()
        target.estimate_normals(search_param=o3d.geometry.KDTreeSearchParamHybrid(radius=0.2,max_nn=10))      
        
        rms = []
        ib = (ia + 1) % len(station_names)
        for b in [station_names[ib]]:
            source = o3d.io.read_point_cloud(os.path.join(path,'{:03d}.txt'.format(b)), format='xyz') 
            # source = o3d.io.read_point_cloud(os.path.join(path,'tank_0_096 cluster_8.txt'),format='xyz')         
            # run ICP
            print("Apply point-to-plane ICP")
            reg_p2l = o3d.pipelines.registration.registration_icp(
                source, target, threshold, trans_init,
                o3d.pipelines.registration.TransformationEstimationPointToPoint(),
                o3d.pipelines.registration.ICPConvergenceCriteria(max_iteration=500))

            # write result
            source_aligned = copy.deepcopy(source).transform(reg_p2l.transformation)
            A = np.asarray(source_aligned.points)
            B = np.asarray(target.points)
            points = np.r_[
                np.c_[ A, np.ones( A.shape[0] )*b ],
                np.c_[ B, np.ones( B.shape[0] )*a ] 
            ]
            # points = np.c_[ A, np.ones( A.shape[0] )*b ]        
            write_array2txt(out_dir + f'\\_aligned_{a}_{b}.txt',points)

            # save transformation matrix
            # with open(out_dir + f'\\transform_{a}{b}.pkl', 'wb') as f:
            #     pickle.dump(reg_p2l.transformation, f)

            # rms.append(reg_p2l.inlier_rmse)
            # print(rms)
            # write_array2txt(out_dir + f'\\rms_{a}{b}.txt',rms,fmt='%.6f')    
```
